# Práctica 4: Multiplicidad, vectores $\alpha$ y $\beta$ y determinación topológica


## 1. Multiplicidad y número de generadores

Trabajamos en el anillo local $\mathcal{O}_3$ (orden `ds`) con el ideal $I = \langle x^5, y^6, z^7, xy^2z^3 \rangle$.

```singular
ring R = 0,(x,y,z),ds;
ideal I = x5,y6,z7,xy2z3;
```

Los cuatro generadores son un sistema generador minimal: $\mu(I) = 4$.

```singular
minbase(I);
```

Calculamos $\dim_{\mathbb{C}} \mathcal{O}_3/I^r$ para $r = 1, \dots, 10$ y sus diferencias sucesivas. La tercera diferencia se estabiliza en la multiplicidad $e(I)$ (se muestra solo esa).

```singular
int r;
intvec H, D1, D2, D3,D4;

for (r = 1; r <= 10; r++)
{
  H[r] = vdim(std(I^r));
}
H;

// primera diferencia: dim I^r / I^(r+1)
for (r = 1; r < size(H); r++)  { D1[r] = H[r+1]  - H[r];  }
D1;

// segunda diferencia
for (r = 1; r < size(D1); r++) { D2[r] = D1[r+1] - D1[r]; }
D2;

// tercera diferencia: se estabiliza en la multiplicidad e(I)
for (r = 1; r < size(D2); r++) { D3[r] = D2[r+1] - D2[r]; }
D3;
```

La multiplicidad es $e(I) = 202$. Otra forma de obtenerla: tres combinaciones lineales genéricas de los generadores dan una reducción $J$ de $I$, que es de intersección completa, así que $e(I) = e(J) = \dim_{\mathbb{C}} \mathcal{O}_3/J$.

```singular
LIB"random.lib";
```

```singular
ideal J = randomid(I,3);
J;
```

```singular
vdim(std(J));
```

Los dos números coinciden. Para estar completamente seguros, $J$ debe cumplir el teorema de la multiplicidad de Rees.

## 2. Reducciones y teorema de Rees

$J = \langle g_1 + g_2, g_3, g_4 \rangle$ no tiene codimensión finita ($-1$): no es reducción.

```singular
ideal J = I[1]+I[2], I[3],I[4];
vdim(std(J));
```

$J = \langle g_1, g_2, g_3 + g_4 \rangle$ tiene $\dim_{\mathbb{C}} \mathcal{O}_3/J = 210 \neq 202$: no es reducción, aunque $J \subseteq I$.

```singular
ideal J = I[1],I[2],I[3]+I[4];
vdim(std(J));
```

```singular
reduce(J,std(I));
```

Con otras tres combinaciones lineales sí obtenemos una reducción. El bucle busca el menor $r$ con $I^{r+1} \subseteq J I^r$ (la otra inclusión es automática): sale $r = 8$.

```singular
ideal J = I[1]+2*I[2]+3*I[3]+5*I[4],
          I[1]-I[2]+4*I[4],
          2*I[2]+I[3]-I[4];
vdim(std(J));

ideal K;
int r = 0;
int found = 0;
while (r <= 10)
{
  K = minbase(reduce(I^(r+1), std(J*I^r)));
  if (K[1] == 0) { found = 1; break; }
  r++;
}

if (found) { "J es reducción de I, r_J(I) =", r; }
else       { "No se cumple hasta r = 10"; }
r;
```

## 3. Los vectores $\alpha$ y $\beta$ y la determinación topológica

### Ejemplo 1

$f = x^2y + xz^3 + x^2y^2z + y^4z + y^3z^2 + x^6 - y^9$ y su ideal jacobiano $J(f)$ (en el código, `I`).

```singular
poly f = x2y +  xz3 + x2y2z + y4z + y3z2 + x6 - y9;
R;
```

```singular
ideal I = jacob(f);
vdim(std(I));
```

```singular
I;
```

Sustituyendo por cero las variables que no queremos mirar vemos las potencias puras de una variable: $\alpha(J(f)) = (2,4,3)$.

```singular
subst(I,x,0,y,0);
```

```singular
subst(I,z,0,y,0);
```

```singular
subst(I,z,0,x,0);
```

Ahora calculamos $\beta$: el menor $r$ con $e(J(f) + \langle x_j^r \rangle) = e(J(f)) = 24$.

```singular
vdim(std(randomid(I+x,3)));
```

```singular
vdim(std(randomid(I+x2,3)));
```

```singular
vdim(std(randomid(I+y3,3)));
```

```singular
vdim(std(randomid(I+y4,3)));
```

```singular
vdim(std(randomid(I+z3,3)));
```

```singular
vdim(std(randomid(I+z4,3)));
```

$\beta(J(f)) = (2,4,4)$ y $\beta_0 = 4$, así que $r_0(f) \in \{4, 5\}$. El 4-jet no tiene singularidad aislada (`milnor` $= -1$), luego no es equivalente a $f$:

```singular
poly g = jet(f,4);
g;
```

```singular
LIB"sing.lib";
milnor(g);
```

```singular
ideal Jg = jacob(g);
vdim(std(Jg));
```

```singular
dim(std(Jg));
```

Por tanto $r_0(f) = 5$; en efecto, $\mu(j^5 f) = 24 = \mu(f)$.

```singular
poly g = jet(f,5);
g;
```

```singular
milnor(g);
```

### Ejemplo 2

$f = xy^4 + x^4z + y^5z + xz^5 + z^7$.

```singular
poly f = xy4 + x4z + y5z + xz5+ z7;
```

```singular
ideal I = jacob(f);
vdim(std(I));
```

$\alpha(J(f)) = (4,4,5)$:

```singular
subst(I,y,0,z,0);
```

```singular
subst(I,x,0,z,0);
```

```singular
subst(I,x,0,y,0);
```

$\beta(J(f)) = (4,5,6)$, luego $\beta_0 = 6$ y $r_0(f) \in \{6, 7\}$:

```singular
vdim(std(randomid(I+x3,3)));
```

```singular
vdim(std(randomid(I+x4,3)));
```

```singular
vdim(std(randomid(I+y4,3)));
```

```singular
vdim(std(randomid(I+y5,3)));
```

```singular
vdim(std(randomid(I+z5,3)));
```

```singular
vdim(std(randomid(I+z6,3)));
```

El 6-jet tiene número de Milnor $85 \neq 84$: no es topológicamente equivalente a $f$, así que $r_0(f) = 7$.

```singular
poly g = jet(f,6);
g;
```

```singular
milnor(g);
```
