## 1. Número de Milnor y número de Tjurina

```singular
ring R=0,(x,y,z),ds;
poly f = xyz2 + x5+y6+z7;
LIB"sing.lib";
milnor(f);
```

El ideal jacobiano `jacob(f)` está generado por las derivadas parciales de orden 1 de $f$.

```singular
ideal Jf=jacob(f);
Jf;
```

Otra forma de calcular el número de Milnor es utilizando `vdim`: $\mu(f) = \dim_{\mathbb{C}} \mathcal{O}_n / J(f)$.

```singular
vdim(std(Jf));
```

El número de Tjurina de $f$ es un invariante analítico: $\tau(f) = \dim_{\mathbb{C}} \dfrac{\mathcal{O}_n}{J(f)+\langle f \rangle} \le \mu(f)$.

```singular
tjurina(f);
```

## 2. Codimensión, dimensión y generadores minimales

$I = \langle g_1, g_2, g_3, g_4 \rangle = \langle xy^2z,\ x^4+yz,\ y^5,\ z^6+y^2x^3 \rangle \subseteq \mathcal{O}_3$.
* `vdim(std(I))`: $\dim_{\mathbb{C}} \mathcal{O}_3/I = 67$.
* `dim(std(I))`: $0$, luego $I$ tiene codimensión finita.
* `minbase(I)`: $\mu(I) = 4$ generadores.

```singular
ideal I = xy2z, x4+yz, y5, z6+y2x3;
vdim(std(I));
```

```singular
dim(std(I));
```

```singular
minbase(I);
```

### Potencias $I^r$

$\dim \mathcal{O}_3/I^r$: $67, 275, 699, 1415, 2499, 4027, 6075$. Diferencias iteradas: $208, 424, 716, 1084, 1528, 2048$; luego $216, 292, 368, 444, 520$; y finalmente $76$ constante: $e(I) = 76 \neq 67$, porque $I$ no es intersección completa ($\mu(I) = 4 > 3$).

```singular
vdim(std(I));
```

```singular
vdim(std(I^2));
```

```singular
vdim(std(I^3));
```

```singular
vdim(std(I^4));
```

```singular
vdim(std(I^5));
```

```singular
vdim(std(I^6));
```

## 3. Teorema del ideal principal de Krull

Si $I$ tiene codimensión finita en $\mathcal{O}_n$, entonces $\mu(I) \ge n$. Este ideal de $\mathcal{O}_4$ tiene $3 < 4$ generadores, luego **no** tiene codimensión finita: `dim` da $1$ y `vdim` da $-1$.

```singular
ring R = 0,(x,y,z,t),ds;
ideal I = x5+xyz, t4+x2, y3+z4;
dim(std(I));
```

```singular
vdim(std(I));
```

## 4. Intersección completa: $e(I) = \dim_{\mathbb{C}} \mathcal{O}_n/I$

Si $I$ tiene codimensión finita y $\mu(I) = n$, entonces $e(I) = \dim_{\mathbb{C}} \mathcal{O}_n/I$. Ejemplo: $I = \langle x^2y+x^7,\ y^6,\ y^2z^2-z^9 \rangle \subseteq \mathcal{O}_3$, con $\dim = 378$.

$\dim \mathcal{O}_3/I^r$: $378, 1512, 3780, 7560, 13230, 21168$, cuya tercera diferencia es $378 = e(I)$.

```singular
ring R = 0,(x,y,z),ds;
```

```singular
ideal I = x2y + x7 , y6, y2z2-z9;
vdim(std(I));
```

```singular
vdim(std(I^2));
```

```singular
vdim(std(I^3));
```

```singular
vdim(std(I^4));
```

```singular
vdim(std(I^5));
```

```singular
vdim(std(I^6));
```

## 5. Reducciones y teorema de Rees

$J \subseteq I$ es una **reducción** de $I$ si $I^{r+1} = J I^r$ para algún $r \ge 0$. **Teorema de Rees:** $e(J) \ge e(I)$, con igualdad si y solo si $J$ es reducción de $I$.

Tres combinaciones lineales genéricas de los $g_i$ (con `randomid`) dan una reducción: $\dim \mathcal{O}_3/J = 76 = e(I)$. (Requiere `LIB"random.lib";`.)

```singular
ideal I = xy2z, x4+yz, y5, z6+y2x3;
ideal J = randomid(I,3);
J;
```

```singular
vdim(std(J));
```

$J = \langle g_1+g_2,\ g_3,\ g_4 \rangle$: $\dim = 120 > 76$, luego **no** es reducción de $I$.

```singular
ideal J = xy2z+ x4+yz, y5, z6+y2x3;
vdim(std(J));
```

$J = \langle g_2,\ g_3+g_1,\ g_4 \rangle$ (`I[k]` es el $k$-ésimo generador): $\dim = 76$, luego **sí** es reducción de $I$.

```singular
ideal J = I[2],I[3]+I[1], I[4];
vdim(std(J));
```

### Comprobación con `reduce`

* `reduce(J, std(I))` $= 0$: $J \subseteq I$.
* `reduce(I, std(J))` $\neq 0$: $I \not\subseteq J$.
* `reduce(I^(r+1), std(J*I^r))` $= 0$ significa $I^{r+1} = J I^r$. Sale no nulo para $r = 1, 2$ y nulo para $r = 3$: el número de reducción es $3$.

```singular
I;
```

```singular
J;
```

```singular
reduce(J,std(I));
```

```singular
reduce(I,std(J));
```

```singular
reduce(I^2,std(J*I));
```

```singular
reduce(I^3,std(J*(I^2)));
```

```singular
reduce(I^4,std(J*(I^3)));
```
