# Práctica 1: Introducción a la geometría analítica compleja y al álgebra conmutativa

El **estudio de subconjuntos analíticos de $\mathbb{C}^n$** significa estudiar el conjunto de ceros de una función holomorfa (en varias variables complejas).

* Sea $U$ un abierto de $\mathbb{C}^n$.
* Sea $f: U \to \mathbb{C}$ función holomorfa $\implies \frac{\partial f}{\partial z_1}, \frac{\partial f}{\partial z_2}, \dots, \frac{\partial f}{\partial z_n}$
* Diremos que $f$ tiene una **singularidad aislada en el origen** cuando $0$ es un punto aislado de:
  $$ \left\{ z \in U \;\middle|\; \frac{\partial f}{\partial z_1}(z) = \dots = \frac{\partial f}{\partial z_n}(z) = 0 \right\} $$

**Ejemplo:**
Sea $f: \mathbb{C}^2 \to \mathbb{C}$ dada por $f(x,y) = x^2y - x^3 + y^4 + x^3y^3 + y^2$.
Dado $r \in \mathbb{Z}_{\ge 1}$, calculamos los **$r$-jets de $f$** ($j^r f$):
* $j^3 f = x^2y - x^3$
* $j^4 f = x^2y - x^3 + y^4$
* $j^2 f = y^2$

## Conceptos preliminares
1. **Anillo:** Sea $A$ un conjunto. Diremos que $(A, +, \cdot)$ es un anillo cuando:
   * Es grupo abeliano con la suma.
   * Cumple la propiedad asociativa: $a \cdot (b \cdot c) = (a \cdot b) \cdot c, \forall a,b,c \in A$.
   * Cumple la propiedad distributiva: $a(b+c) = ab+ac$.
   * **Anillo conmutativo:** $ab=ba, \forall a,b \in A$.
   * **Anillo unitario:** cuando $\exists 1_A$ tal que $1_A \cdot a = a \cdot 1_A, \forall a \in A$.
   * *Ejemplo:* $\mathbb{C}[x_1, \dots, x_n]$ (No tiene dimensión finita).
2. **Subanillo:** $B$ es subanillo de $A$ cuando... (definición análoga heredada).
3. **Cuerpo:** Sea $A$ un anillo. Decimos que $A$ es cuerpo cuando $A \setminus \{0\}$ es un grupo abeliano.
   * *Ejemplo:* $\mathbb{C}, \mathbb{Z}_p = \mathbb{Z}/p\mathbb{Z}$ (con $p$ primo).
4. **Característica:** Sea $k$ cuerpo. La característica de $k$ es $char(k) = \min \{n \in \mathbb{Z}_{\ge 1} \mid n \cdot 1_k = 0_k\}$. Si no existe tal $n$, entonces $char(k)=0$.
   * *Ejemplos:* $char(\mathbb{C})=char(\mathbb{R})=0$, $char(\mathbb{Z}_p) = p$ (con $p$ primo).

## Ideales
Sean $I, J$ ideales de un anillo $A$.

* **Suma de ideales:**
  $$ I+J = \{a+b \mid a \in I, b \in J\} $$
  Si $I = \langle g_1, \dots, g_r \rangle$ y $J = \langle h_1, \dots, h_s \rangle \implies I+J = \langle g_1, \dots, g_r, h_1, \dots, h_s \rangle$
* **Producto de ideales:**
  $$ I \cdot J = \left\{ \sum_{j=1}^l \alpha_j \beta_j \;\middle|\; \alpha_j \in I, \beta_j \in J, \forall j \right\} = \langle g_i h_j \mid i=1,\dots,r; j=1,\dots,s \rangle $$
* **Intersección de ideales:**
  $$ I \cap J = \{a \mid a \in I, a \in J\} $$

Se cumple la relación de inclusión: $I \cdot J \subseteq I \cap J \subseteq I+J$.

## Homomorfismo de anillos
Sea $f: R \to S$ aplicación entre anillos. Diremos que $f$ es un homomorfismo de anillos cuando:
* a) $f(1_R) = 1_S$
* b) $f(a+_R b) = f(a) +_S f(b) \quad \forall a,b \in R$
* c) $f(a \cdot_R b) = f(a) \cdot_S f(b) \quad \forall a,b \in R$

**Ejemplo:**
Sea la aplicación $F: \mathbb{C}[a,b,c] \to \mathbb{C}[x,y,z]$ (donde el primer anillo, $S$, es $\mathbb{C}[a,b,c]$ y el segundo, $R$, es $\mathbb{C}[x,y,z]$).
Se define sobre los generadores:
$F(a) = x^2, \quad F(b) = y^3, \quad F(c) = x^4 - 2y^5$
Y $F$ se extiende de manera lineal. Por ejemplo, si tenemos el polinomio $p(a,b,c) = a^3 - b^2c$, su imagen será:
$$ F(p(a,b,c)) = F(a)^3 - F(b)^2 F(c) $$

Singular nos permite calcular con rapidez estos elementos:
Utiliza Bases de Grobner para eso.

```singular
ring R = 0, (x,y,z),dp ;
ideal I = x3y - xy4z , x+y-z;
ideal J = 3x+5y, 7y-z2 ;
ideal S = I+J;
ideal P = I*J;
ideal U = intersect(I,J);
```

# Homomorfismo de anillos


Se puede definir un homomorfismo a traves de la definición del generador y la extensión lineal

```singular
ring S = 0, (a,b,c), dp;
ring R = 0, (x,y,z), dp;
ideal I  = x2, y3, x4-2y5;
map F = S,I;
F;
```

Comprobamos si es inyectiva, monomorfismo, miramos si $ker(F) = \{p \in S : F(p)=0\} = \{0\}$

```singular
LIB"algebra.lib";
```

```singular
is_injective(F,S);
```

```singular
alg_kernel(F,S);
```

Si intentamos definirlo como ideal falla puesto que el kernel es ideal de S y no de R

```singular
ideal G = alg_kernel(F,S);
```

```singular
is_surjective(F,S);
```

```singular
is_bijective(F,S);
```
